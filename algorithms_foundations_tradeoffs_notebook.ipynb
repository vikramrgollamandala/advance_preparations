# 2. Breadth-First Search vs Depth-First Search

Both BFS and DFS traverse graphs or trees, but they prioritize nodes differently.

## Breadth-First Search (BFS)

Idea:
- explore neighbors level by level
- usually implemented with a queue

Typical runtime:
- `O(V + E)` for graph traversal

Typical memory:
- can be large because it stores an entire frontier

Strengths:
- finds shortest path in an unweighted graph
- useful for level-based exploration

Weaknesses:
- high memory on wide graphs

# 3. Divide and Conquer

Idea:
- split a problem into smaller subproblems
- solve subproblems
- combine results

Typical strengths:
- enables more efficient algorithms than simple scanning in many cases
- often works well with sorting and searching

Typical limitations:
- recursive overhead
- sometimes requires sorted data or expensive combine steps
